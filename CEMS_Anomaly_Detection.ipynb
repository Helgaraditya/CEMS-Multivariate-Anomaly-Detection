* Pipeline Library

```python
def isolation_forest_pipeline(X_window, contamination_list):

    results = {}
    summary = []

    for c in contamination_list:

        print(f"\n=== Contamination: {c} ===")

        # 1. Inisialisasi model
        model = IsolationForest(
            n_estimators=300,
            contamination=c,
            random_state=42
        )

        # 2. Training
        model.fit(X_window)

        # 3. Anomaly scoring
        scores = model.decision_function(X_window)

        # 4. Threshold dari percentile
        threshold = np.percentile(scores, 100 * c)

        # 5. Labeling otomatis
        labels_model = model.predict(X_window)

        # 6. Labeling manual
        labels_manual = np.where(scores < threshold, -1, 1)

        # 7. Hitung anomaly
        anomaly_count = np.sum(labels_model == -1)

        # 8. Validasi
        same = np.sum(labels_model == labels_manual)

        print("Jumlah anomaly:", anomaly_count)
        print("Threshold:", threshold)
        print("Label cocok:", same, "/", len(X_window))

        # 9. Simpan hasil
        results[c] = {
            'model': model,
            'scores': scores,
            'labels_model': labels_model,
            'labels_manual': labels_manual,
            'threshold': threshold,
            'anomaly_count': anomaly_count
        }

        # 10. Summary
        summary.append({
            'contamination': c,
            'jumlah_anomali': anomaly_count,
            'threshold': threshold
        })

    summary_df = pd.DataFrame(summary)

    return results, summary_df
```

Step 1: Hitung anomaly score =  merepresentasikan tingkat keanehan data, di mana nilai yang lebih rendah menunjukkan data yang lebih anomali

Step 2: Mengurutkan score

Step 3: menentukan proporsi anomali  = 5% data dianggap anomali

Step 4 : Hitung percentile
* 100 × 0.05 = 5
* threshold = nilai pada 5% data terbawah
* nilai score yang memisahkan X% data paling ekstrem dari sisanya

## Rentang Eksklusif (Range Classification)

Visualisasi menunjukkan hasil deteksi anomali harian yang dibagi menjadi tiga tingkat keparahan berdasarkan jumlah jam anomali dalam satu hari, yaitu Minor (1–5 anomaly/day), Moderate (6–10 anomaly/day), dan Severe (≥11 anomaly/day).

1. Minor Anomaly (1–5 anomaly/day)

Terdapat 11 hari (3.01%) yang termasuk kategori minor.

Hal ini menunjukkan bahwa pada beberapa hari terjadi penyimpangan ringan yang bersifat sementara atau sesaat. Anomali minor umumnya muncul akibat fluktuasi proses normal, perubahan beban operasi, atau noise sensor, sehingga belum menunjukkan gangguan operasional yang serius.

Makna:
* Gangguan kecil dan sporadis
* Masih dalam kondisi relatif terkendali
* Perlu pemantauan, tetapi belum kritis

---

2. Moderate Anomaly (6–10 anomaly/day)

Terdapat 7 hari (1.92%) yang termasuk kategori moderate.

Kategori ini menunjukkan gangguan yang lebih nyata karena kondisi abnormal berlangsung selama beberapa jam dalam satu hari. Moderate anomaly mengindikasikan proses mulai tidak stabil dan dapat berdampak pada kualitas operasi maupun emisi.

Makna:
* Gangguan proses tingkat sedang
* Terjadi selama seperempat hingga hampir setengah hari
* Perlu perhatian operator dan evaluasi penyebab

---

3. Severe Anomaly (≥11 anomaly/day)

Terdapat 19 hari (5.21%) yang masuk kategori severe.

Jumlah ini lebih tinggi dibanding minor, yang menunjukkan bahwa ketika anomali terjadi, sebagian besar cenderung berlangsung cukup lama dalam satu hari. Titik anomali banyak muncul pada periode pertengahan hingga akhir tahun, terutama saat beberapa variabel mengalami perubahan ekstrem secara bersamaan.

Makna:
* Indikasi gangguan proses yang signifikan
* Ketidakstabilan operasi berlangsung berjam-jam
* Berpotensi terkait shutdown parsial, perubahan proses besar, atau deviasi emisi


Anomali tidak hanya terjadi pada satu variabel, tetapi muncul bersamaan pada beberapa variabel. Hal ini menunjukkan bahwa gangguan yang terjadi bersifat multivariat, yaitu mempengaruhi sistem operasi secara keseluruhan, bukan hanya satu sensor.

Mayoritas hari pengamatan berada dalam kondisi normal, namun ketika anomali muncul, kecenderungannya lebih sering berada pada level severe dibanding minor. Ini menandakan bahwa sistem umumnya stabil, tetapi ketika terjadi gangguan, dampaknya cukup signifikan dan berlangsung lebih lama.

Berdasarkan hasil visualisasi, sebagian besar hari pengamatan berada pada kondisi normal. Namun demikian, ditemukan 11 hari anomali minor (3.01%) dan 19 hari anomali severe (5.21%). Hal ini menunjukkan bahwa meskipun frekuensi anomali relatif rendah, gangguan yang terjadi cenderung memiliki tingkat keparahan tinggi dan mempengaruhi beberapa variabel proses secara simultan.


```python
import matplotlib.pyplot as plt
import pandas as pd

# ==========================================
# DAILY DATA
# ==========================================
df_daily = df_plot.drop(columns='anomaly').resample('D').mean()
daily_count = (df_plot['anomaly'] == -1).resample('D').sum()

# ==========================================
# LEVEL ANOMALY
# ==========================================
minor_days = daily_count[(daily_count >= 1) & (daily_count <= 5)].index
moderate_days = daily_count[(daily_count >= 6) & (daily_count <= 10)].index
severe_days = daily_count[daily_count >= 11].index

# ==========================================
# JUMLAH & PERSENTASE
# ==========================================
total_days = len(daily_count)

minor_n = len(minor_days)
moderate_n = len(moderate_days)
severe_n = len(severe_days)

minor_pct = minor_n / total_days * 100
moderate_pct = moderate_n / total_days * 100
severe_pct = severe_n / total_days * 100


# =====================================================
# FUNCTION PLOT
# =====================================================
def plot_level(days, color, title, count, pct):

    plt.figure(figsize=(18,7))

    # --------------------------------------
    # Plot tiap variabel + label legend
    # --------------------------------------
    for col in df_daily.columns:
        plt.plot(
            df_daily.index,
            df_daily[col],
            linewidth=1.4,
            alpha=0.8,
            label=col
        )

        plt.scatter(
            days,
            df_daily.loc[days, col],
            color=color,
            s=25,
            alpha=0.9
        )

    # --------------------------------------
    # Dummy scatter utk legend anomaly
    # --------------------------------------
    plt.scatter([], [], color=color, s=35, label='Anomaly Day')

    # --------------------------------------
    # Title
    # --------------------------------------
    plt.title(
        f"{title}\nJumlah Hari = {count} ({pct:.2f}%)",
        fontsize=16
    )

    plt.xlabel("Date")
    plt.ylabel("Average Standardized Value")

    # --------------------------------------
    # Legend
    # --------------------------------------
    plt.legend(
        loc='upper right',
        ncol=2,
        fontsize=9,
        frameon=True
    )

    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()


# ==========================================
# 1. MINOR
# ==========================================
plot_level(
    minor_days,
    color='red',
    title="Minor Anomaly (1–5 anomaly/day)",
    count=minor_n,
    pct=minor_pct
)

# ==========================================
# 2. MODERATE
# ==========================================
plot_level(
    moderate_days,
    color='red',
    title="Moderate Anomaly (6–10 anomaly/day)",
    count=moderate_n,
    pct=moderate_pct
)

# ==========================================
# 3. SEVERE
# ==========================================
plot_level(
    severe_days,
    color='red',
    title="Severe Anomaly (≥11 anomaly/day)",
    count=severe_n,
    pct=severe_pct
)
```

## PIPELINE LOF

```python
# =====================================================
# 0) IMPORT
# =====================================================
import numpy as np
import pandas as pd
from sklearn.neighbors import LocalOutlierFactor, NearestNeighbors

# =====================================================
# 1) VALIDASI DATA
# =====================================================
print("Shape window_features:", X_window.shape)

# sinkronisasi index
time_index_aligned = df_plot.index[-len(X_window):]

print("Length time_index:", len(time_index_aligned))

assert len(time_index_aligned) == X_window.shape[0], "Index tidak sesuai!"

# =====================================================
# 2) PARAMETER MODEL
# =====================================================
k_neighbors = 30

contamination_primary = 0.02
contamination_secondary = 0.05

# =====================================================
# 3) KNN (UNTUK EXPLAINABILITY)
# =====================================================
knn_model = NearestNeighbors(n_neighbors=k_neighbors)
knn_model.fit(X_window)

dist_matrix, neighbor_idx_matrix = knn_model.kneighbors(X_window)

# k-distance
k_dist_vector = dist_matrix[:, -1]

# =====================================================
# 4) REACHABILITY DISTANCE
# =====================================================
reach_dist_matrix = np.maximum(dist_matrix, k_dist_vector[:, None])

# =====================================================
# 5) LOCAL REACHABILITY DENSITY (LRD)
# =====================================================
lrd_vector = 1 / (np.mean(reach_dist_matrix, axis=1) + 1e-10)

# =====================================================
# 6) LOF SCORE (MANUAL - ANALISIS)
# =====================================================
lof_score_manual = []

for i in range(len(X_window)):
    neighbors = neighbor_idx_matrix[i]
    ratio = lrd_vector[neighbors] / lrd_vector[i]
    lof_score_manual.append(np.mean(ratio))

lof_score_manual = np.array(lof_score_manual)

# =====================================================
# 7) FUNCTION MODEL LOF
# =====================================================
def run_lof_model(data_input, time_index, k_neighbors, contamination_rate):

    lof_estimator = LocalOutlierFactor(
        n_neighbors=k_neighbors,
        contamination=contamination_rate
    )

    label_output = lof_estimator.fit_predict(data_input)
    lof_score_model = -lof_estimator.negative_outlier_factor_

    df_output = pd.DataFrame({
        'Score_LOF': lof_score_model,
        'Label_LOF_Model': label_output
    }, index=time_index)

    df_output['Flag_Anomaly_Model'] = (df_output['Label_LOF_Model'] == -1).astype(int)

    return df_output

# =====================================================
# 8) RUN MODEL (2 KONFIGURASI)
# =====================================================
df_lof_primary = run_lof_model(
    X_window,
    time_index_aligned,
    k_neighbors,
    contamination_primary
)

df_lof_secondary = run_lof_model(
    X_window,
    time_index_aligned,
    k_neighbors,
    contamination_secondary
)

# =====================================================
# 9) LABEL MANUAL BERDASARKAN PERCENTILE
# =====================================================
def generate_percentile_label(df_input, percentile_value):

    threshold_value = np.percentile(df_input['Score_LOF'], percentile_value)

    df_input[f'Threshold_P{percentile_value}'] = threshold_value
    df_input[f'Flag_Anomaly_P{percentile_value}'] = (
        df_input['Score_LOF'] > threshold_value
    ).astype(int)

    return df_input, threshold_value

# mapping:
# 98% ≈ contamination 0.02
# 95% ≈ contamination 0.05

df_lof_primary, threshold_p98 = generate_percentile_label(df_lof_primary, 98)
df_lof_secondary, threshold_p95 = generate_percentile_label(df_lof_secondary, 95)

print("Threshold P98:", threshold_p98)
print("Threshold P95:", threshold_p95)

# =====================================================
# 10) VALIDASI KECOCOKAN LABEL
# =====================================================
def evaluate_label_match(df_input, model_col, manual_col):

    total_match = (df_input[model_col] == df_input[manual_col]).sum()
    total_data = len(df_input)

    match_percentage = (total_match / total_data) * 100

    print(f"\nPerbandingan {model_col} vs {manual_col}")
    print("Total Match :", total_match)
    print("Total Data  :", total_data)
    print("Match (%)   :", round(match_percentage, 2))

evaluate_label_match(df_lof_primary, 'Flag_Anomaly_Model', 'Flag_Anomaly_P98')
evaluate_label_match(df_lof_secondary, 'Flag_Anomaly_Model', 'Flag_Anomaly_P95')

# =====================================================
# 11) RINGKASAN HASIL
# =====================================================
print("\n=== LOF PRIMARY (0.02) ===")
print("Model anomaly :", df_lof_primary['Flag_Anomaly_Model'].sum())
print("Manual anomaly:", df_lof_primary['Flag_Anomaly_P98'].sum())

print("\n=== LOF SECONDARY (0.05) ===")
print("Model anomaly :", df_lof_secondary['Flag_Anomaly_Model'].sum())
print("Manual anomaly:", df_lof_secondary['Flag_Anomaly_P95'].sum())

# =====================================================
# 12) CEK OUTPUT
# =====================================================
print("\nSample Output:")
print(df_lof_primary.head())
```

## Range Classification Comparison Daily IF vs LOF

```python
# =====================================================
# IF vs LOF DAILY ANOMALY SEVERITY COMPARISON
# MULTIVARIATE DAILY VISUALIZATION
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# =====================================================
# 1) MULTIVARIATE VARIABLES
# =====================================================

multi_vars = [
    'Particulate (mg/Nm3)',
    'SO2 (mg/Nm3)',
    'NOx (mg/Nm3)',
    'CO (mg/Nm3)',
    'O2 ( % )',
    'Flowrate (m/s)'
]

# =====================================================
# 2) DAILY MEAN MULTIVARIATE
# =====================================================

daily_multivar = (
    df_plot[multi_vars]
    .resample('D')
    .mean()
)

# =====================================================
# 3) DAILY ANOMALY COUNT
# =====================================================

# ---------------------------------------------
# IF
# -1 = anomaly
# ---------------------------------------------
daily_if_count = (
    (df_plot['Anomaly_IF_05'] == -1)
    .astype(int)
    .resample('D')
    .sum()
)

# ---------------------------------------------
# LOF
# 1 = anomaly
# ---------------------------------------------
daily_lof_count = (
    (df_plot['Anomaly_LOF_05'] == 1)
    .astype(int)
    .resample('D')
    .sum()
)

# =====================================================
# 4) SEVERITY FILTER FUNCTION
# =====================================================

def severity_filter(series, level):

    if level == 'Early Warning':
        return series[(series >= 1) & (series <= 5)]

    elif level == 'Process Disturbance':
        return series[(series >= 6) & (series <= 10)]

    elif level == 'Critical Operating Condition':
        return series[(series >= 11)]

# =====================================================
# 5) VISUALIZATION FUNCTION
# =====================================================

def plot_severity_comparison(level_name):

    # ---------------------------------------------
    # FILTER DAILY COUNT
    # ---------------------------------------------
    if_event = severity_filter(
        daily_if_count,
        level_name
    )

    lof_event = severity_filter(
        daily_lof_count,
        level_name
    )

    # ---------------------------------------------
    # EVENT DAYS
    # ---------------------------------------------
    if_days = if_event.index
    lof_days = lof_event.index

    # ---------------------------------------------
    # TOTAL EVENT
    # ---------------------------------------------
    total_days = len(daily_multivar)

    total_if = len(if_days)
    total_lof = len(lof_days)

    pct_if = (total_if / total_days) * 100
    pct_lof = (total_lof / total_days) * 100

    # =================================================
    # FIGURE
    # =================================================
    plt.figure(figsize=(24,10))

    # ---------------------------------------------
    # MULTIVARIATE DAILY TREND
    # ---------------------------------------------
    for col in multi_vars:

        plt.plot(
            daily_multivar.index,
            daily_multivar[col],
            linewidth=1.5,
            alpha=0.8,
            label=col
        )

    # ---------------------------------------------
    # IF ANOMALY
    # merah bulat
    # ---------------------------------------------
    for i, col in enumerate(multi_vars):

        plt.scatter(
            if_days,
            daily_multivar.loc[if_days, col],
            color='red',
            marker='o',
            s=100,
            linewidths=1.0,
            alpha=0.8,
            label='IF Anomaly' if i == 0 else ""
        )

    # ---------------------------------------------
    # LOF ANOMALY
    # hitam silang
    # ---------------------------------------------
    for i, col in enumerate(multi_vars):

        plt.scatter(
            lof_days,
            daily_multivar.loc[lof_days, col],
            color='black',
            marker='x',
            s=120,
            linewidths=2.0,
            alpha=0.8,
            label='LOF Anomaly' if i == 0 else ""
        )

    # =================================================
    # TITLE
    # =================================================

    if level_name == 'Early Warning':

        subtitle = (
            '(1–5 anomaly/hour per day)'
        )

    elif level_name == 'Process Disturbance':

        subtitle = (
            '(6–10 anomaly/hour per day)'
        )

    else:

        subtitle = (
            '(≥11 anomaly/hour per day)'
        )

    plt.title(
        #f'IF vs LOF Daily Anomaly Severity Comparison — {level_name}\n'
        #f'{subtitle}\n'
        f'IF : Total Hari = {total_if} ({pct_if:.2f}%) | '
        f'LOF : Total Hari = {total_lof} ({pct_lof:.2f}%)',
        fontsize=20,
        fontweight='bold'
    )

    # =================================================
    # AXIS
    # =================================================

    plt.xlabel(
        'Time',
        fontsize=14
    )

    plt.ylabel(
        'Daily Mean Standardized Value',
        fontsize=14
    )

    # =================================================
    # X-AXIS FORMAT
    # =================================================

    ax = plt.gca()

    ax.xaxis.set_major_locator(
        mdates.MonthLocator(interval=1)
    )

    ax.xaxis.set_major_formatter(
        mdates.DateFormatter('%Y-%m')
    )

    plt.xticks(rotation=45)

    # =================================================
    # GRID
    # =================================================

    plt.grid(alpha=0.3)

    # =================================================
    # LEGEND
    # =================================================

    plt.legend(
        fontsize=10,
        ncol=2,
        loc='upper right'
    )

    # =================================================
    # LAYOUT
    # =================================================

    plt.tight_layout()

    # =================================================
    # SHOW
    # =================================================

    plt.show()

    # =================================================
    # PRINT EVENT TABLE
    # =================================================

    severity_table = pd.DataFrame({
        'IF_Daily_Count': if_event,
        'LOF_Daily_Count': lof_event
    })

    print("\n=================================================")
    print(f"{level_name.upper()} EVENT")
    print("=================================================")

    print(severity_table)

# =====================================================
# 6) EARLY WARNING
# =====================================================

plot_severity_comparison(
    'Early Warning'
)

# =====================================================
# 7) PROCESS DISTURBANCE
# =====================================================

plot_severity_comparison(
    'Process Disturbance'
)

# =====================================================
# 8) CRITICAL OPERATING CONDITION
# =====================================================

plot_severity_comparison(
    'Critical Operating Condition'
)
```

```python
# =====================================================
# IF vs LOF DAILY ANOMALY SEVERITY COMPARISON
# MULTIVARIATE DAILY VISUALIZATION
# =====================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

# =====================================================
# 1) MULTIVARIATE VARIABLES
# =====================================================

multi_vars = [
    'Particulate (mg/Nm3)',
    'SO2 (mg/Nm3)',
    'NOx (mg/Nm3)',
    'CO (mg/Nm3)',
    'O2 ( % )',
    'Flowrate (m/s)'
]

# =====================================================
# 2) DAILY MEAN MULTIVARIATE
# =====================================================

daily_multivar = (
    df_plot[multi_vars]
    .resample('D')
    .mean()
)

# =====================================================
# 3) DAILY ANOMALY COUNT
# =====================================================

# ---------------------------------------------
# IF
# -1 = anomaly
# ---------------------------------------------
daily_if_count = (
    (df_plot['Anomaly_IF_05'] == -1)
    .astype(int)
    .resample('D')
    .sum()
)

# ---------------------------------------------
# LOF
# 1 = anomaly
# ---------------------------------------------
daily_lof_count = (
    (df_plot['Anomaly_LOF_05'] == 1)
    .astype(int)
    .resample('D')
    .sum()
)

# =====================================================
# 4) SEVERITY FILTER FUNCTION
# =====================================================

def severity_filter(series, level):

    if level == 'Early Warning':
        return series[(series >= 1) & (series <= 5)]

    elif level == 'Process Disturbance':
        return series[(series >= 6) & (series <= 10)]

    elif level == 'Critical Operating Condition':
        return series[(series >= 11)]

# =====================================================
# 5) VISUALIZATION FUNCTION
# =====================================================

def plot_severity_comparison(level_name):

    # ---------------------------------------------
    # FILTER DAILY COUNT
    # ---------------------------------------------
    if_event = severity_filter(
        daily_if_count,
        level_name
    )

    lof_event = severity_filter(
        daily_lof_count,
        level_name
    )

    # ---------------------------------------------
    # EVENT DAYS
    # ---------------------------------------------
    if_days = if_event.index
    lof_days = lof_event.index

    # ---------------------------------------------
    # TOTAL EVENT
    # ---------------------------------------------
    total_days = len(daily_multivar)

    total_if = len(if_days)
    total_lof = len(lof_days)

    pct_if = (total_if / total_days) * 100
    pct_lof = (total_lof / total_days) * 100

    # =================================================
    # FIGURE
    # =================================================
    plt.figure(figsize=(24,10))

    # ---------------------------------------------
    # MULTIVARIATE DAILY TREND
    # ---------------------------------------------
    for col in multi_vars:

        plt.plot(
            daily_multivar.index,
            daily_multivar[col],
            linewidth=1.5,
            alpha=0.8,
            label=col
        )

    # ---------------------------------------------
    # IF ANOMALY
    # merah bulat
    # ---------------------------------------------
    for i, col in enumerate(multi_vars):

        plt.scatter(
            if_days,
            daily_multivar.loc[if_days, col],
            color='red',
            marker='o',
            s=28,
            alpha=0.8,
            label='IF Anomaly' if i == 0 else ""
        )

    # ---------------------------------------------
    # LOF ANOMALY
    # hitam silang
    # ---------------------------------------------
    for i, col in enumerate(multi_vars):

        plt.scatter(
            lof_days,
            daily_multivar.loc[lof_days, col],
            color='black',
            marker='x',
            s=28,
            alpha=0.8,
            label='LOF Anomaly' if i == 0 else ""
        )

    # =================================================
    # TITLE
    # =================================================

    if level_name == 'Early Warning':

        subtitle = (
            '(1–5 anomaly/hour per day)'
        )

    elif level_name == 'Process Disturbance':

        subtitle = (
            '(6–10 anomaly/hour per day)'
        )

    else:

        subtitle = (
            '(≥11 anomaly/hour per day)'
        )

    plt.title(
        #f'IF vs LOF Daily Anomaly Severity Comparison — {level_name}\n'
        #f'{subtitle}\n'
        f'IF : Total Day = {total_if} ({pct_if:.2f}%) | '
        f'LOF : Total Day = {total_lof} ({pct_lof:.2f}%)',
        fontsize=20,
        fontweight='bold'
    )

    # =================================================
    # AXIS
    # =================================================

    plt.xlabel(
        'Time',
        fontsize=14
    )

    plt.ylabel(
        'Daily Mean Standardized Value',
        fontsize=14
    )

    # =================================================
    # X-AXIS FORMAT
    # =================================================

    ax = plt.gca()

    ax.xaxis.set_major_locator(
        mdates.MonthLocator(interval=1)
    )

    ax.xaxis.set_major_formatter(
        mdates.DateFormatter('%Y-%m')
    )

    plt.xticks(rotation=45)

    # =================================================
    # GRID
    # =================================================

    plt.grid(alpha=0.3)

    # =================================================
    # LEGEND
    # =================================================

    plt.legend(
        fontsize=10,
        ncol=2,
        loc='upper right'
    )

    # =================================================
    # LAYOUT
    # =================================================

    plt.tight_layout()

    # =================================================
    # SHOW
    # =================================================

    plt.show()

    # =================================================
    # PRINT EVENT TABLE
    # =================================================

    severity_table = pd.DataFrame({
        'IF_Daily_Count': if_event,
        'LOF_Daily_Count': lof_event
    })

    print("\n=================================================")
    print(f"{level_name.upper()} EVENT")
    print("=================================================")

    print(severity_table)

# =====================================================
# 6) EARLY WARNING
# =====================================================

plot_severity_comparison(
    'Early Warning'
)

# =====================================================
# 7) PROCESS DISTURBANCE
# =====================================================

plot_severity_comparison(
    'Process Disturbance'
)

# =====================================================
# 8) CRITICAL OPERATING CONDITION
# =====================================================

plot_severity_comparison(
    'Critical Operating Condition'
)
```

```python
# =========================================================
# DOMINANT VARIABLE ANALYSIS DURING ANOMALY
# =========================================================
#
# TUJUAN:
# Mengetahui variabel mana yang paling dominan berubah
# saat anomaly terjadi.
#
# METODE:
# Membandingkan rata-rata absolut nilai standar
# saat anomaly.
#
# Semakin besar nilai absolut:
# -> semakin besar deviasi dari kondisi normal
# -> semakin dominan variabel tersebut
#
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =========================================================
# VARIABLES
# =========================================================

variables = [
    'Particulate (mg/Nm3)',
    'SO2 (mg/Nm3)',
    'NOx (mg/Nm3)',
    'CO (mg/Nm3)',
    'O2 ( % )',
    'Flowrate (m/s)'
]

# =========================================================
# 1) AMBIL DATA ANOMALY
# =========================================================

# ---------------------------------------------------------
# IF ANOMALY
# ---------------------------------------------------------
# IF :
# -1 = anomaly
# ---------------------------------------------------------

df_if_anomaly = df_plot[
    df_plot['Anomaly_IF_05'] == -1
]

# ---------------------------------------------------------
# LOF ANOMALY
# ---------------------------------------------------------
# LOF :
# 1 = anomaly
# ---------------------------------------------------------

df_lof_anomaly = df_plot[
    df_plot['Anomaly_LOF_05'] == 1
]

# =========================================================
# 2) HITUNG ABSOLUTE MEAN
# =========================================================
#
# Menggunakan nilai absolut karena:
# anomaly bisa spike (+)
# atau drop (-)
#
# Semakin besar nilai absolut:
# semakin dominan perubahan variabel
#
# =========================================================

# ---------------------------------------------------------
# IF
# ---------------------------------------------------------

if_dominant = (
    df_if_anomaly[variables]
    .abs()
    .mean()
    .sort_values(ascending=False)
)

# ---------------------------------------------------------
# LOF
# ---------------------------------------------------------

lof_dominant = (
    df_lof_anomaly[variables]
    .abs()
    .mean()
    .sort_values(ascending=False)
)

# =========================================================
# 3) SUMMARY TABLE
# =========================================================

df_dominant = pd.DataFrame({

    'IF_Mean_Abs': if_dominant,

    'LOF_Mean_Abs': lof_dominant
})

# =========================================================
# 4) DISPLAY SUMMARY
# =========================================================

print("\n================================================")
print("DOMINANT VARIABLE DURING ANOMALY")
print("================================================\n")

print(df_dominant)

# =========================================================
# 5) VISUALIZATION
# =========================================================

x = np.arange(len(df_dominant.index))

width = 0.35

plt.figure(figsize=(14,7))

# ---------------------------------------------------------
# IF BAR
# ---------------------------------------------------------

plt.bar(
    x - width/2,
    df_dominant['IF_Mean_Abs'],
    width,
    label='IF'
)

# ---------------------------------------------------------
# LOF BAR
# ---------------------------------------------------------

plt.bar(
    x + width/2,
    df_dominant['LOF_Mean_Abs'],
    width,
    label='LOF'
)

# =========================================================
# 6) LABEL
# =========================================================

plt.xticks(
    x,
    df_dominant.index,
    rotation=15,
    fontsize=11
)

plt.ylabel(
    'Mean Absolute Standardized Value',
    fontsize=13
)

plt.xlabel(
    'Variables',
    fontsize=13
)

# =========================================================
# 7) TITLE
# =========================================================

plt.title(
    'Dominant Variable During Anomaly\n'
    'IF vs LOF Comparison',
    fontsize=18,
    fontweight='bold'
)

# =========================================================
# 8) GRID
# =========================================================

plt.grid(
    axis='y',
    alpha=0.3
)

# =========================================================
# 9) LEGEND
# =========================================================

plt.legend(
    fontsize=12
)

# =========================================================
# 10) VALUE LABEL
# =========================================================

for i, value in enumerate(df_dominant['IF_Mean_Abs']):

    plt.text(
        i - width/2,
        value + 0.02,
        f'{value:.2f}',
        ha='center',
        fontsize=10
    )

for i, value in enumerate(df_dominant['LOF_Mean_Abs']):

    plt.text(
        i + width/2,
        value + 0.02,
        f'{value:.2f}',
        ha='center',
        fontsize=10
    )

# =========================================================
# 11) LAYOUT
# =========================================================

plt.tight_layout()

# =========================================================
# 12) SHOW
# =========================================================

plt.show()

# =========================================================
# 13) TOP DOMINANT VARIABLE
# =========================================================

print("\n================================================")
print("TOP DOMINANT VARIABLE")
print("================================================\n")

print(
    f"IF Dominant Variable  : "
    f"{if_dominant.index[0]}"
)

print(
    f"LOF Dominant Variable : "
    f"{lof_dominant.index[0]}"
)

# =========================================================
# 14) INTERPRETATION
# =========================================================

print("\n================================================")
print("INTERPRETATION")
print("================================================\n")

print(
    "- Nilai mean absolute yang lebih besar "
    "menunjukkan variabel lebih dominan berubah "
    "saat anomaly."
)

print(
    "- Variabel dominan merepresentasikan "
    "indikator utama disturbance sistem."
)

print(
    "- IF cenderung menangkap perubahan global "
    "yang lebih besar."
)

print(
    "- LOF cenderung sensitif terhadap "
    "perubahan lokal multivariat."
)

# =========================================================
# 15) OPTIONAL OPERATIONAL INSIGHT
# =========================================================

print("\n================================================")
print("OPERATIONAL INSIGHT")
print("================================================\n")

# ---------------------------------------------------------
# CO
# ---------------------------------------------------------

if 'CO (mg/Nm3)' in df_dominant.index[:3]:

    print(
        "- CO dominan dapat mengindikasikan "
        "incomplete combustion."
    )

# ---------------------------------------------------------
# O2
# ---------------------------------------------------------

if 'O2 ( % )' in df_dominant.index[:3]:

    print(
        "- O2 dominan menunjukkan perubahan "
        "kondisi pembakaran atau suplai udara."
    )

# ---------------------------------------------------------
# NOx
# ---------------------------------------------------------

if 'NOx (mg/Nm3)' in df_dominant.index[:3]:

    print(
        "- NOx dominan dapat berkaitan dengan "
        "temperatur pembakaran tinggi."
    )

# ---------------------------------------------------------
# SO2
# ---------------------------------------------------------

if 'SO2 (mg/Nm3)' in df_dominant.index[:3]:

    print(
        "- SO2 dominan dapat mengindikasikan "
        "gangguan sulfur/material proses."
    )

# ---------------------------------------------------------
# Flowrate
# ---------------------------------------------------------

if 'Flowrate (m/s)' in df_dominant.index[:3]:

    print(
        "- Flowrate dominan menunjukkan "
        "perubahan aliran gas sistem."
    )

# ---------------------------------------------------------
# Particulate
# ---------------------------------------------------------

if 'PParticulate (mg/Nm3)' in df_dominant.index[:3]:

    print(
        "- Particulate dominan dapat berkaitan "
        "dengan dust carry-over atau filter disturbance."
    )
```

```python
# =========================================================
# UNSUPERVISED MODEL EVALUATION
# IF vs LOF ANOMALY DETECTION
# CEMS MULTIVARIATE EVALUATION
# =========================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# =========================================================
# VARIABLES
# =========================================================

variables = [
    'Particulate (mg/Nm3)',
    'SO2 (mg/Nm3)',
    'NOx (mg/Nm3)',
    'CO (mg/Nm3)',
    'O2 ( % )',
    'Flowrate (m/s)'
]

# =========================================================
# 1) ANOMALY FLAG
# =========================================================

# ---------------------------------------------------------
# IF
# -1 = anomaly
# ---------------------------------------------------------

df_plot['IF_Flag'] = np.where(
    df_plot['Anomaly_IF_05'] == -1,
    1,
    0
)

# ---------------------------------------------------------
# LOF
# 1 = anomaly
# ---------------------------------------------------------

df_plot['LOF_Flag'] = np.where(
    df_plot['Anomaly_LOF_05'] == 1,
    1,
    0
)

# =========================================================
# 2) OVERLAP FLAG
# =========================================================

df_plot['Overlap_Flag'] = np.where(
    (df_plot['IF_Flag'] == 1) &
    (df_plot['LOF_Flag'] == 1),
    1,
    0
)

# =========================================================
# 3) TOTAL ANOMALY
# =========================================================

total_if = df_plot['IF_Flag'].sum()
total_lof = df_plot['LOF_Flag'].sum()
total_overlap = df_plot['Overlap_Flag'].sum()

# =========================================================
# 4) PERCENTAGE
# =========================================================

total_data = len(df_plot)

pct_if = (total_if / total_data) * 100
pct_lof = (total_lof / total_data) * 100
pct_overlap = (total_overlap / total_data) * 100

# =========================================================
# 5) DOMINANT VARIABLE ANALYSIS
# =========================================================

# ---------------------------------------------------------
# IF
# ---------------------------------------------------------

if_mean_abs = (
    df_plot[df_plot['IF_Flag'] == 1][variables]
    .abs()
    .mean()
)

# ---------------------------------------------------------
# LOF
# ---------------------------------------------------------

lof_mean_abs = (
    df_plot[df_plot['LOF_Flag'] == 1][variables]
    .abs()
    .mean()
)

# =========================================================
# 6) CORRELATION ANALYSIS
# =========================================================

# ---------------------------------------------------------
# IF
# ---------------------------------------------------------

corr_if = (
    df_plot[df_plot['IF_Flag'] == 1][variables]
    .corr()
)

# ---------------------------------------------------------
# LOF
# ---------------------------------------------------------

corr_lof = (
    df_plot[df_plot['LOF_Flag'] == 1][variables]
    .corr()
)

# =========================================================
# 7) DAILY CLUSTER ANALYSIS
# =========================================================

# ---------------------------------------------------------
# DAILY IF COUNT
# ---------------------------------------------------------

daily_if = (
    df_plot['IF_Flag']
    .resample('D')
    .sum()
)

# ---------------------------------------------------------
# DAILY LOF COUNT
# ---------------------------------------------------------

daily_lof = (
    df_plot['LOF_Flag']
    .resample('D')
    .sum()
)

# ---------------------------------------------------------
# CLUSTER DAY
# >5 anomaly/hour
# ---------------------------------------------------------

cluster_if = (
    daily_if[daily_if > 5]
)

cluster_lof = (
    daily_lof[daily_lof > 5]
)

# =========================================================
# 8) SUMMARY TABLE
# =========================================================

summary_eval = pd.DataFrame({

    'Evaluation Aspect': [

        'Total Anomaly',
        'Anomaly Percentage',
        'Overlap Anomaly',
        'Cluster Days (>5 anomaly/day)',
        'Most Dominant Variable',
        'Strongest Correlation',
        'Detection Characteristic'

    ],

    'IF': [

        total_if,
        f'{pct_if:.2f}%',
        total_overlap,
        len(cluster_if),
        if_mean_abs.idxmax(),
        (
            corr_if.where(
                np.triu(
                    np.ones(corr_if.shape),
                    k=1
                ).astype(bool)
            )
            .stack()
            .idxmax()
        ),
        'Global / Stable'

    ],

    'LOF': [

        total_lof,
        f'{pct_lof:.2f}%',
        total_overlap,
        len(cluster_lof),
        lof_mean_abs.idxmax(),
        (
            corr_lof.where(
                np.triu(
                    np.ones(corr_lof.shape),
                    k=1
                ).astype(bool)
            )
            .stack()
            .idxmax()
        ),
        'Local / Sensitive'

    ]
})

# =========================================================
# 9) DISPLAY SUMMARY
# =========================================================

print("\n================================================")
print("UNSUPERVISED MODEL EVALUATION SUMMARY")
print("================================================\n")

print(summary_eval)

# =========================================================
# 10) VISUALIZATION
# =========================================================

fig, axes = plt.subplots(
    2,
    2,
    figsize=(18,12)
)

# =========================================================
# 11) TOTAL ANOMALY
# =========================================================

axes[0,0].bar(
    ['IF', 'LOF'],
    [total_if, total_lof]
)

axes[0,0].set_title(
    'Total Detected Anomaly',
    fontsize=14,
    fontweight='bold'
)

axes[0,0].set_ylabel('Total Hourly Anomaly')

# =========================================================
# 12) OVERLAP
# =========================================================

axes[0,1].bar(
    ['Overlap'],
    [total_overlap]
)

axes[0,1].set_title(
    'Overlap Anomaly',
    fontsize=14,
    fontweight='bold'
)

axes[0,1].set_ylabel('Total Overlap')

# =========================================================
# 13) DOMINANT VARIABLE
# =========================================================

x = np.arange(len(variables))
width = 0.35

axes[1,0].bar(
    x - width/2,
    if_mean_abs.values,
    width,
    label='IF'
)

axes[1,0].bar(
    x + width/2,
    lof_mean_abs.values,
    width,
    label='LOF'
)

axes[1,0].set_xticks(x)
axes[1,0].set_xticklabels(
    variables,
    rotation=15
)

axes[1,0].set_title(
    'Dominant Variable During Anomaly',
    fontsize=14,
    fontweight='bold'
)

axes[1,0].legend()

# =========================================================
# 14) DAILY CLUSTER
# =========================================================

axes[1,1].plot(
    daily_if.index,
    daily_if.values,
    label='IF'
)

axes[1,1].plot(
    daily_lof.index,
    daily_lof.values,
    label='LOF'
)

axes[1,1].axhline(
    y=5,
    linestyle='--',
    label='Cluster Threshold'
)

axes[1,1].set_title(
    'Daily Anomaly Cluster',
    fontsize=14,
    fontweight='bold'
)

axes[1,1].legend()

# =========================================================
# 15) LAYOUT
# =========================================================

plt.tight_layout()

# =========================================================
# 16) SHOW
# =========================================================

plt.show()

# =========================================================
# 17) CORRELATION HEATMAP
# =========================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(18,7)
)

# ---------------------------------------------------------
# IF
# ---------------------------------------------------------

sns.heatmap(
    corr_if,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    fmt='.2f',
    ax=axes[0]
)

axes[0].set_title(
    'IF Correlation During Anomaly',
    fontsize=15,
    fontweight='bold'
)

# ---------------------------------------------------------
# LOF
# ---------------------------------------------------------

sns.heatmap(
    corr_lof,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1,
    fmt='.2f',
    ax=axes[1]
)

axes[1].set_title(
    'LOF Correlation During Anomaly',
    fontsize=15,
    fontweight='bold'
)

plt.tight_layout()
plt.show()

# =========================================================
# 18) INTERPRETATION
# =========================================================

print("\n================================================")
print("MODEL EVALUATION INTERPRETATION")
print("================================================\n")

# ---------------------------------------------------------
# TOTAL ANOMALY
# ---------------------------------------------------------

if total_lof > total_if:

    print(
        "- LOF mendeteksi anomaly lebih banyak "
        "yang menunjukkan sensitivitas lebih tinggi."
    )

else:

    print(
        "- IF mendeteksi anomaly lebih banyak "
        "yang menunjukkan kemampuan global isolation."
    )

# ---------------------------------------------------------
# OVERLAP
# ---------------------------------------------------------

print(
    "- Overlap anomaly menunjukkan periode "
    "anomali yang terdeteksi secara konsisten "
    "oleh kedua metode."
)

# ---------------------------------------------------------
# CLUSTER
# ---------------------------------------------------------

print(
    "- Cluster anomaly menunjukkan bahwa anomaly "
    "tidak muncul secara random tetapi membentuk "
    "gangguan temporal persisten."
)

# ---------------------------------------------------------
# MULTIVARIATE
# ---------------------------------------------------------

print(
    "- Korelasi antar variabel menunjukkan "
    "bahwa anomaly bersifat multivariat."
)

# ---------------------------------------------------------
# OPERATIONAL ANALYSIS
# ---------------------------------------------------------

print(
    "- Variabel dominan seperti NOx, CO, O2, "
    "dan SO2 mengindikasikan hubungan anomaly "
    "dengan proses pembakaran dan emisi."
)

# ---------------------------------------------------------
# METHOD CHARACTER
# ---------------------------------------------------------

print(
    "- IF cenderung mendeteksi anomaly global "
    "dan lebih stabil."
)

print(
    "- LOF lebih sensitif terhadap "
    "fluktuasi lokal."
)

# =========================================================
# 19) FINAL CONCLUSION
# =========================================================

print("\n================================================")
print("FINAL EVALUATION CONCLUSION")
print("================================================\n")

print(
    "Evaluasi model unsupervised dilakukan "
    "menggunakan pendekatan berbasis "
    "konsistensi temporal, overlap anomaly, "
    "hubungan multivariat, dan interpretasi "
    "operasional sistem CEMS."
)

print()

print(
    "Hasil menunjukkan bahwa anomaly yang "
    "terdeteksi membentuk cluster temporal "
    "dan melibatkan perubahan beberapa "
    "variabel secara simultan."
)

print()

print(
    "Isolation Forest cenderung mendeteksi "
    "disturbance global dengan deviasi besar "
    "dan lebih stabil, sedangkan LOF lebih "
    "sensitif terhadap fluktuasi lokal."
)

print()

print(
    "Dengan demikian, kedua metode mampu "
    "merepresentasikan perilaku anomaly "
    "multivariat pada sistem CEMS secara baik."
)
```

## 1 . Import, konfigurasi, dan fungsi umum

```python
# ============================================================
# 1. IMPORT, KONFIGURASI, DAN FUNGSI UMUM
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from itertools import combinations
from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor


# ------------------------------------------------------------
# DATA INPUT
# ------------------------------------------------------------

X = np.asarray(X_window, dtype=float)


# ------------------------------------------------------------
# VALIDASI DATA
# ------------------------------------------------------------

if X.ndim != 2:
    raise ValueError(
        "X_window harus berbentuk 2 dimensi: "
        "(jumlah_observasi, jumlah_fitur)."
    )

if not np.isfinite(X).all():
    raise ValueError(
        "X_window masih mengandung NaN atau infinite value."
    )

print("Shape X_window :", X.shape)


# ------------------------------------------------------------
# GRID PARAMETER
# ------------------------------------------------------------

N_ESTIMATORS_LIST = [50, 100, 200, 300]
N_NEIGHBORS_LIST = [10, 20, 30, 40]

CONTAMINATION_LIST = [0.01, 0.02, 0.03, 0.05, 0.07]

# Digunakan hanya sebagai baseline untuk memilih
# n_estimators dan n_neighbors.
BASE_CONTAMINATION = 0.05

RANDOM_STATES = [11, 22, 33, 42, 55]

N_LOF_RUNS = 5
LOF_NOISE_SCALE = 0.001
LOF_RANDOM_STATE = 42

FINAL_RANDOM_STATE = 42


# ------------------------------------------------------------
# MENGAMBIL INDEKS ANOMALI
# ------------------------------------------------------------

def get_anomaly_set(labels, anomaly_label=-1):
    """
    Mengubah label model menjadi himpunan indeks anomali.
    """

    labels = np.asarray(labels)

    return set(
        np.where(labels == anomaly_label)[0]
    )


# ------------------------------------------------------------
# JACCARD SIMILARITY
# ------------------------------------------------------------

def jaccard_similarity(set_a, set_b):
    """
    Menghitung Jaccard Similarity:

        J(A, B) = |A ∩ B| / |A ∪ B|

    Nilai mendekati 1 menunjukkan kesamaan yang tinggi.
    """

    set_a = set(set_a)
    set_b = set(set_b)

    union = set_a | set_b

    if len(union) == 0:
        return 1.0

    intersection = set_a & set_b

    return len(intersection) / len(union)


# ------------------------------------------------------------
# RATA-RATA PAIRWISE JACCARD
# ------------------------------------------------------------

def calculate_stability(anomaly_sets):
    """
    Menghitung rata-rata dan standar deviasi Jaccard
    dari seluruh pasangan hasil deteksi.
    """

    if len(anomaly_sets) < 2:
        return np.nan, np.nan

    scores = [
        jaccard_similarity(set_a, set_b)
        for set_a, set_b in combinations(anomaly_sets, 2)
    ]

    return float(np.mean(scores)), float(np.std(scores))
```

## 2 . Stability Analysis (IF)

```python
# ============================================================
# 2. STABILITY ANALYSIS - ISOLATION FOREST
# ============================================================

def run_if_stability_analysis(
    X,
    n_estimators_list,
    contamination,
    random_states
):
    """
    Menguji stabilitas setiap nilai n_estimators
    menggunakan beberapa random_state.

    Parameter terbaik dipilih berdasarkan:
    1. mean_jaccard tertinggi;
    2. std_jaccard terendah;
    3. jika masih sama, n_estimators lebih kecil.
    """

    result_rows = []

    for n_estimators in n_estimators_list:

        anomaly_sets = []
        anomaly_counts = []

        for seed in random_states:

            model = IsolationForest(
                n_estimators=n_estimators,
                contamination=contamination,
                random_state=seed,
                n_jobs=-1
            )

            labels = model.fit_predict(X)

            current_anomaly_set = get_anomaly_set(labels)

            anomaly_sets.append(current_anomaly_set)
            anomaly_counts.append(len(current_anomaly_set))

        mean_jaccard, std_jaccard = calculate_stability(
            anomaly_sets
        )

        result_rows.append({
            "n_estimators": n_estimators,
            "contamination": contamination,
            "mean_jaccard": mean_jaccard,
            "std_jaccard": std_jaccard,
            "mean_anomaly_count": np.mean(anomaly_counts),
            "min_anomaly_count": np.min(anomaly_counts),
            "max_anomaly_count": np.max(anomaly_counts)
        })

    result_df = pd.DataFrame(result_rows)

    ranked_df = result_df.sort_values(
        by=[
            "mean_jaccard",
            "std_jaccard",
            "n_estimators"
        ],
        ascending=[
            False,
            True,
            True
        ]
    ).reset_index(drop=True)

    best_n_estimators = int(
        ranked_df.loc[0, "n_estimators"]
    )

    return (
        result_df.sort_values("n_estimators").reset_index(drop=True),
        best_n_estimators
    )


if_stability_result, best_n_estimators = (
    run_if_stability_analysis(
        X=X,
        n_estimators_list=N_ESTIMATORS_LIST,
        contamination=BASE_CONTAMINATION,
        random_states=RANDOM_STATES
    )
)


print("\n" + "=" * 70)
print("STABILITY ANALYSIS - ISOLATION FOREST")
print("=" * 70)

print(
    if_stability_result.to_string(index=False)
)

print(
    "\nBest n_estimators berdasarkan stability:",
    best_n_estimators
)
```

## 3 . Stability Analysis (LOF)

```python
# ============================================================
# 3. STABILITY ANALYSIS - LOCAL OUTLIER FACTOR
# ============================================================

def generate_perturbed_datasets(
    X,
    n_runs,
    noise_scale,
    random_state
):
    """
    Membentuk beberapa data perturbasi dengan noise kecil.
    """

    rng = np.random.default_rng(random_state)

    perturbed_datasets = []

    for _ in range(n_runs):

        noise = rng.normal(
            loc=0.0,
            scale=noise_scale,
            size=X.shape
        )

        perturbed_datasets.append(
            X + noise
        )

    return perturbed_datasets


lof_perturbed_datasets = generate_perturbed_datasets(
    X=X,
    n_runs=N_LOF_RUNS,
    noise_scale=LOF_NOISE_SCALE,
    random_state=LOF_RANDOM_STATE
)


def run_lof_stability_analysis(
    X,
    n_neighbors_list,
    contamination,
    perturbed_datasets
):
    """
    Menguji stabilitas n_neighbors LOF terhadap
    perubahan kecil pada data.

    Parameter terbaik dipilih berdasarkan:
    1. mean_jaccard tertinggi;
    2. std_jaccard terendah;
    3. jika masih sama, n_neighbors lebih kecil.
    """

    result_rows = []

    for n_neighbors in n_neighbors_list:

        if n_neighbors >= X.shape[0]:
            raise ValueError(
                f"n_neighbors={n_neighbors} harus lebih kecil "
                f"daripada jumlah observasi={X.shape[0]}."
            )

        anomaly_sets = []
        anomaly_counts = []

        for X_perturbed in perturbed_datasets:

            model = LocalOutlierFactor(
                n_neighbors=n_neighbors,
                contamination=contamination,
                novelty=False,
                n_jobs=-1
            )

            labels = model.fit_predict(X_perturbed)

            current_anomaly_set = get_anomaly_set(labels)

            anomaly_sets.append(current_anomaly_set)
            anomaly_counts.append(len(current_anomaly_set))

        mean_jaccard, std_jaccard = calculate_stability(
            anomaly_sets
        )

        result_rows.append({
            "n_neighbors": n_neighbors,
            "contamination": contamination,
            "mean_jaccard": mean_jaccard,
            "std_jaccard": std_jaccard,
            "mean_anomaly_count": np.mean(anomaly_counts),
            "min_anomaly_count": np.min(anomaly_counts),
            "max_anomaly_count": np.max(anomaly_counts)
        })

    result_df = pd.DataFrame(result_rows)

    ranked_df = result_df.sort_values(
        by=[
            "mean_jaccard",
            "std_jaccard",
            "n_neighbors"
        ],
        ascending=[
            False,
            True,
            True
        ]
    ).reset_index(drop=True)

    best_n_neighbors = int(
        ranked_df.loc[0, "n_neighbors"]
    )

    return (
        result_df.sort_values("n_neighbors").reset_index(drop=True),
        best_n_neighbors
    )


lof_stability_result, best_n_neighbors = (
    run_lof_stability_analysis(
        X=X,
        n_neighbors_list=N_NEIGHBORS_LIST,
        contamination=BASE_CONTAMINATION,
        perturbed_datasets=lof_perturbed_datasets
    )
)


print("\n" + "=" * 70)
print("STABILITY ANALYSIS - LOCAL OUTLIER FACTOR")
print("=" * 70)

print(
    lof_stability_result.to_string(index=False)
)

print(
    "\nBest n_neighbors berdasarkan stability:",
    best_n_neighbors
)
```

## 4 . Fungsi Stability Setiap Contamination

```python
# ============================================================
# 4. FUNGSI STABILITY UNTUK CONTAMINATION
# ============================================================

def calculate_if_contamination_stability(
    X,
    n_estimators,
    contamination,
    random_states
):
    """
    Menghitung stability IF pada satu nilai contamination.
    """

    anomaly_sets = []

    for seed in random_states:

        model = IsolationForest(
            n_estimators=n_estimators,
            contamination=contamination,
            random_state=seed,
            n_jobs=-1
        )

        labels = model.fit_predict(X)

        anomaly_sets.append(
            get_anomaly_set(labels)
        )

    return calculate_stability(anomaly_sets)


def calculate_lof_contamination_stability(
    n_neighbors,
    contamination,
    perturbed_datasets
):
    """
    Menghitung stability LOF pada satu nilai contamination.
    """

    anomaly_sets = []

    for X_perturbed in perturbed_datasets:

        model = LocalOutlierFactor(
            n_neighbors=n_neighbors,
            contamination=contamination,
            novelty=False,
            n_jobs=-1
        )

        labels = model.fit_predict(X_perturbed)

        anomaly_sets.append(
            get_anomaly_set(labels)
        )

    return calculate_stability(anomaly_sets)
```

# Contamination Evaluation

```python
# ============================================================
# 5. CONTAMINATION EVALUATION
# ============================================================

def run_contamination_evaluation(
    X,
    contamination_list,
    n_estimators,
    n_neighbors,
    random_states,
    perturbed_datasets,
    final_random_state=42
):
    """
    Mengevaluasi contamination menggunakan:

    1. IF stability
    2. LOF stability
    3. Jaccard overlap IF–LOF
    4. Jumlah anomali
    5. IF Only, LOF Only, dan Both Detected

    Fungsi tidak memilih contamination secara otomatis.
    """

    result_rows = []
    label_storage = {}

    for contamination in contamination_list:

        # ----------------------------------------------------
        # ISOLATION FOREST
        # ----------------------------------------------------

        if_model = IsolationForest(
            n_estimators=n_estimators,
            contamination=contamination,
            random_state=final_random_state,
            n_jobs=-1
        )

        if_labels = if_model.fit_predict(X)
        if_set = get_anomaly_set(if_labels)

        # ----------------------------------------------------
        # LOCAL OUTLIER FACTOR
        # ----------------------------------------------------

        lof_model = LocalOutlierFactor(
            n_neighbors=n_neighbors,
            contamination=contamination,
            novelty=False,
            n_jobs=-1
        )

        lof_labels = lof_model.fit_predict(X)
        lof_set = get_anomaly_set(lof_labels)

        # ----------------------------------------------------
        # OVERLAP
        # ----------------------------------------------------

        both_set = if_set & lof_set
        union_set = if_set | lof_set

        if_only_set = if_set - lof_set
        lof_only_set = lof_set - if_set

        overlap_jaccard = jaccard_similarity(
            if_set,
            lof_set
        )

        # ----------------------------------------------------
        # STABILITY IF
        # ----------------------------------------------------

        if_stability_mean, if_stability_std = (
            calculate_if_contamination_stability(
                X=X,
                n_estimators=n_estimators,
                contamination=contamination,
                random_states=random_states
            )
        )

        # ----------------------------------------------------
        # STABILITY LOF
        # ----------------------------------------------------

        lof_stability_mean, lof_stability_std = (
            calculate_lof_contamination_stability(
                n_neighbors=n_neighbors,
                contamination=contamination,
                perturbed_datasets=perturbed_datasets
            )
        )

        # ----------------------------------------------------
        # PERCENTAGE
        # ----------------------------------------------------

        both_vs_if_pct = (
            100 * len(both_set) / len(if_set)
            if len(if_set) > 0 else 0
        )

        both_vs_lof_pct = (
            100 * len(both_set) / len(lof_set)
            if len(lof_set) > 0 else 0
        )

        both_vs_union_pct = (
            100 * len(both_set) / len(union_set)
            if len(union_set) > 0 else 0
        )

        # ----------------------------------------------------
        # SIMPAN HASIL
        # ----------------------------------------------------

        result_rows.append({
            "contamination": contamination,

            "if_stability": if_stability_mean,
            "if_stability_std": if_stability_std,

            "lof_stability": lof_stability_mean,
            "lof_stability_std": lof_stability_std,

            "overlap_jaccard": overlap_jaccard,

            "if_anomaly": len(if_set),
            "lof_anomaly": len(lof_set),

            "if_only": len(if_only_set),
            "lof_only": len(lof_only_set),
            "both_detected": len(both_set),
            "union_anomaly": len(union_set),

            "both_vs_if_pct": both_vs_if_pct,
            "both_vs_lof_pct": both_vs_lof_pct,
            "both_vs_union_pct": both_vs_union_pct
        })

        label_storage[contamination] = {
            "IF_Label": if_labels,
            "LOF_Label": lof_labels,
            "IF_Anomaly_Index": sorted(if_set),
            "LOF_Anomaly_Index": sorted(lof_set),
            "IF_Only_Index": sorted(if_only_set),
            "LOF_Only_Index": sorted(lof_only_set),
            "Both_Index": sorted(both_set)
        }

    result_df = pd.DataFrame(result_rows)

    result_df = result_df.sort_values(
        "contamination"
    ).reset_index(drop=True)

    return result_df, label_storage


contamination_result, contamination_storage = (
    run_contamination_evaluation(
        X=X,
        contamination_list=CONTAMINATION_LIST,
        n_estimators=best_n_estimators,
        n_neighbors=best_n_neighbors,
        random_states=RANDOM_STATES,
        perturbed_datasets=lof_perturbed_datasets,
        final_random_state=FINAL_RANDOM_STATE
    )
)


display_columns = [
    "contamination",
    "if_stability",
    "lof_stability",
    "overlap_jaccard",
    "if_anomaly",
    "lof_anomaly",
    "both_detected",
    "union_anomaly"
]


print("\n" + "=" * 90)
print("CONTAMINATION EVALUATION")
print("=" * 90)

print(
    contamination_result[
        display_columns
    ].to_string(index=False)
)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

# =========================================================
# 1. COPY DATA
# =========================================================

df_overlap = df_plot.copy()

# Pastikan index berupa DatetimeIndex
if not isinstance(df_overlap.index, pd.DatetimeIndex):
    if 'DATETIME' in df_overlap.columns:
        df_overlap['DATETIME'] = pd.to_datetime(
            df_overlap['DATETIME'],
            errors='coerce'
        )
        df_overlap = df_overlap.set_index('DATETIME')
    else:
        raise ValueError(
            "Data harus memiliki DatetimeIndex atau kolom 'DATETIME'."
        )

df_overlap = df_overlap.sort_index()


# =========================================================
# 2. VALIDATE REQUIRED COLUMNS
# =========================================================

if_column = 'Anomaly_IF_05'
lof_column = 'Anomaly_LOF_05'

required_columns = [if_column, lof_column]

missing_columns = [
    column for column in required_columns
    if column not in df_overlap.columns
]

if missing_columns:
    raise KeyError(
        f"Kolom label tidak ditemukan: {missing_columns}"
    )


# =========================================================
# 3. CREATE BINARY ANOMALY FLAGS
# =========================================================

# Isolation Forest:
# -1 = anomaly
#  1 = normal
df_overlap['IF_Flag'] = (
    df_overlap[if_column]
    .eq(-1)
    .astype(int)
)

# Local Outlier Factor:
# 1 = anomaly
# -1 = normal
df_overlap['LOF_Flag'] = (
    df_overlap[lof_column]
    .eq(1)
    .astype(int)
)


# =========================================================
# 4. CREATE OVERLAP CATEGORIES
# =========================================================

df_overlap['Overlap_Category'] = 'Normal'

df_overlap.loc[
    (df_overlap['IF_Flag'] == 1) &
    (df_overlap['LOF_Flag'] == 0),
    'Overlap_Category'
] = 'IF Only'

df_overlap.loc[
    (df_overlap['IF_Flag'] == 0) &
    (df_overlap['LOF_Flag'] == 1),
    'Overlap_Category'
] = 'LOF Only'

df_overlap.loc[
    (df_overlap['IF_Flag'] == 1) &
    (df_overlap['LOF_Flag'] == 1),
    'Overlap_Category'
] = 'Both Detected'


# =========================================================
# 5. DISPLAY OVERLAP SUMMARY
# =========================================================

overlap_summary = (
    df_overlap['Overlap_Category']
    .value_counts()
    .reindex(
        ['Normal', 'IF Only', 'LOF Only', 'Both Detected'],
        fill_value=0
    )
)

total_if_only = int(overlap_summary['IF Only'])
total_lof_only = int(overlap_summary['LOF Only'])
total_both = int(overlap_summary['Both Detected'])
total_union = total_if_only + total_lof_only + total_both

jaccard_overlap = (
    total_both / total_union
    if total_union > 0
    else 0
)

print("Overlap Summary")
print("-" * 40)
print(f"IF Only       : {total_if_only}")
print(f"LOF Only      : {total_lof_only}")
print(f"Both Detected : {total_both}")
print(f"Union Anomaly : {total_union}")
print(f"Jaccard       : {jaccard_overlap:.4f}")


# =========================================================
# 6. DEFINE VARIABLES
# =========================================================

variables = [
    'Particulate (mg/Nm3)',
    'SO2 (mg/Nm3)',
    'NOx (mg/Nm3)',
    'CO (mg/Nm3)',
    'O2 ( % )',
    'Flowrate (m/s)'
]

# Sesuaikan nama berikut jika nama kolom pada data berbeda
missing_variables = [
    variable for variable in variables
    if variable not in df_overlap.columns
]

if missing_variables:
    raise KeyError(
        f"Kolom variabel tidak ditemukan: {missing_variables}"
    )


# =========================================================
# 7. CREATE MASKS
# =========================================================

mask_if_only = (
    df_overlap['Overlap_Category'] == 'IF Only'
)

mask_lof_only = (
    df_overlap['Overlap_Category'] == 'LOF Only'
)

mask_both = (
    df_overlap['Overlap_Category'] == 'Both Detected'
)


# =========================================================
# 8. CREATE SUBPLOT 3 x 2
# =========================================================

fig, axes = plt.subplots(
    nrows=3,
    ncols=2,
    figsize=(20, 14),
    sharex=True
)

axes = axes.flatten()

for ax, variable in zip(axes, variables):

    # -----------------------------------------
    # Full time-series as background
    # -----------------------------------------
    ax.plot(
        df_overlap.index,
        df_overlap[variable],
        color='lightgray',
        linewidth=0.65,
        alpha=0.85,
        zorder=1
    )

    # -----------------------------------------
    # IF Only
    # -----------------------------------------
    ax.scatter(
        df_overlap.index[mask_if_only],
        df_overlap.loc[mask_if_only, variable],
        color='red',
        s=14,
        alpha=0.80,
        marker='o',
        edgecolors='none',
        zorder=3
    )

    # -----------------------------------------
    # LOF Only
    # -----------------------------------------
    ax.scatter(
        df_overlap.index[mask_lof_only],
        df_overlap.loc[mask_lof_only, variable],
        color='orange',
        s=14,
        alpha=0.80,
        marker='^',
        edgecolors='none',
        zorder=3
    )

    # -----------------------------------------
    # Both Detected
    # -----------------------------------------
    ax.scatter(
        df_overlap.index[mask_both],
        df_overlap.loc[mask_both, variable],
        color='green',
        s=24,
        alpha=0.95,
        marker='X',
        edgecolors='black',
        linewidths=0.25,
        zorder=4
    )

    ax.set_title(
        variable,
        fontsize=11,
        fontweight='bold'
    )

    ax.set_ylabel('Standardized Value')
    ax.grid(
        True,
        alpha=0.25,
        linewidth=0.6
    )


# =========================================================
# 9. X-AXIS LABEL
# =========================================================

for ax in axes[-2:]:
    ax.set_xlabel('Date')


# =========================================================
# 10. CUSTOM LEGEND
# =========================================================

legend_elements = [
    Line2D(
        [0],
        [0],
        color='lightgray',
        linewidth=1.5,
        label='All Observations'
    ),
    Line2D(
        [0],
        [0],
        marker='o',
        color='none',
        markerfacecolor='red',
        markeredgecolor='none',
        markersize=7,
        label=f'IF Only ({total_if_only})'
    ),
    Line2D(
        [0],
        [0],
        marker='^',
        color='none',
        markerfacecolor='orange',
        markeredgecolor='none',
        markersize=7,
        label=f'LOF Only ({total_lof_only})'
    ),
    Line2D(
        [0],
        [0],
        marker='X',
        color='none',
        markerfacecolor='green',
        markeredgecolor='black',
        markeredgewidth=0.4,
        markersize=8,
        label=f'Both Detected ({total_both})'
    )
]

fig.legend(
    handles=legend_elements,
    loc='upper center',
    bbox_to_anchor=(0.5, 0.955),
    ncol=4,
    frameon=True,
    fontsize=10
)


# =========================================================
# 11. MAIN TITLE
# =========================================================

fig.suptitle(
    (
        'Long-Term Overlap Anomaly Detection: '
        'Isolation Forest and Local Outlier Factor\n'
        f'IF Only = {total_if_only} | '
        f'LOF Only = {total_lof_only} | '
        f'Both Detected = {total_both} | '
        f'Jaccard = {jaccard_overlap:.4f}'
    ),
    fontsize=15,
    fontweight='bold',
    y=0.995
)


# =========================================================
# 12. FORMAT AND DISPLAY
# =========================================================

fig.autofmt_xdate(
    rotation=30,
    ha='right'
)

plt.tight_layout(
    rect=[0.03, 0.03, 0.98, 0.91]
)

plt.show()
```
